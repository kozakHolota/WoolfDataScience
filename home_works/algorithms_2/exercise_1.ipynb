```python
def get_k_lowest_element(array: list, k: int) -> int:
    """
    Find the k-th smallest element in an unsorted array using the Quick Select algorithm.

    The function selects a pivot element and partitions the array into elements
    smaller than, equal to, and greater than the pivot. It then continues the search
    only in the partition that can contain the k-th smallest element, avoiding a full
    sort of the array.

    Args:
        array (list[int | float]): A non-empty list of numeric values.
        k (int): The 1-based position of the element to find.
            For example, k=1 returns the smallest element,
            k=2 returns the second smallest element, and so on.

    Returns:
        int | float: The k-th smallest element in the array.
    """
    pivot = array[len(array) // 2]
    left: list = []
    right: list = []
    for elem in array:
        if elem <= pivot:
            left.append(elem)
        else:
            right.append(elem)

    return left[k - 1] if k <= len(left) else right[(k - 1) - len(left)]
```

```python

# Example usage of get_k_lowest_element()

data = [7, 2, 10, 1, 5, 3, 8]

print("Array:", data)
print()

# Find 1st, 2nd, 3rd smallest elements
for k in range(1, 4):
    result = get_k_lowest_element(data, k)
    print(f"k={k} → {k}-th smallest element: {result}")

print()

# Edge cases
single = [42]
print("Single-element array:", get_k_lowest_element(single, 1))  # 42

negatives = [-5, -1, -9, -3]
print("Negatives, k=2:", get_k_lowest_element(negatives, 2))  # -5

floats = [3.14, 1.41, 2.71, 0.57]
print("Floats, k=1 (smallest):", get_k_lowest_element(floats, 1))  # 0.57
```
