```python
from category_encoders import TargetEncoder
from pandas import read_csv

df = read_csv("ml-fundamentals-and-applications-2025-10/final_proj_data.csv")
df.head()
```

```python
num_cols = df.select_dtypes(include=["number"]).columns.tolist()
num_cols
```

```python
cat_cols = df.drop(columns=num_cols).columns.tolist()
cat_cols
```

```python
from home_works.machine_learning.final_project.util import get_nan_percentage

allowed_columns = df.loc[:, get_nan_percentage(df) < 35].columns.tolist()

df_cleaned = df[allowed_columns]
df_cleaned.head()
```

```python
from home_works.machine_learning.final_project.util import split_df_to_x_y

X, y = split_df_to_x_y(df_cleaned, "y")
```

```python
from sklearn.pipeline import Pipeline
from home_works.machine_learning.final_project.util import VarMeanEstimator
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

original_num_cols = num_cols
num_cols = [c for c in num_cols if c in X.columns]

num_data_prep = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("var_mean", VarMeanEstimator(col_name="VarMean", cols_to_mean=num_cols))
])

num_data_prep.set_output(transform="pandas")

num_data_scaled = num_data_prep.fit_transform(X[num_cols])

num_data_scaled.head()
```

```python
from home_works.machine_learning.final_project.util import CategoryHasher

original_cat_cols = cat_cols
cat_cols = [c for c in original_cat_cols if c in X.columns]

cat_prep = Pipeline(
    steps=[
        ("target_encoder", TargetEncoder(cols=cat_cols)),
    ]
)

cat_prep.set_output(transform="pandas")

cat_preped = cat_prep.fit_transform(X[cat_cols], y)

cat_preped.head()
```

```python
from home_works.machine_learning.final_project.util import concat_dfs

X_final = concat_dfs(num_data_scaled, cat_preped)
X_final.head()
```

```python
from home_works.machine_learning.final_project.util import HyperParametersOptimizer

best_params = HyperParametersOptimizer(X_final, y).catboost_objective(n_trials=100)

best_params
```

```python
from imblearn.ensemble import BalancedRandomForestClassifier

model = BalancedRandomForestClassifier(**best_params)
```

```python
from home_works.machine_learning.final_project.util import evaluate_model_avg_accuracy

evaluate_model_avg_accuracy(model, X_final, y)
```

```python
from home_works.machine_learning.final_project.util import evaluate_model, split_train_test

X_train, X_test, y_train, y_test = split_train_test(X_final, y)
evaluate_model(model, X_train, y_train, X_test, y_test)
```

```python
final_model = BalancedRandomForestClassifier(**best_params)
final_model.fit(X_final, y)
```

```python
validation_data = read_csv("ml-fundamentals-and-applications-2025-10/final_proj_test.csv")

validation_data.head()
```

```python
allowed_columns.remove("y")
df_cleaned_final = validation_data[allowed_columns]
df_cleaned_final.head()
```

```python
num_cols_final = df_cleaned_final.select_dtypes(include=["number"]).columns.tolist()

df_cleaned_final[num_cols_final].head()
```

```python
cat_cols_final = df_cleaned_final.drop(columns=num_cols_final).columns.tolist()
df_cleaned_final[cat_cols_final].head()
```

```python
# Використовуємо .transform(), а не .fit_transform().
# Пайплайни вже "навчені" на X та y вище. Тепер ми просто застосовуємо знання до нових даних.

# 1. Числові дані: Тільки transform
num_data_scaled_final = num_data_prep.transform(df_cleaned_final[num_cols_final])
num_data_scaled_final.head()
```

```python
# 2. Категоріальні дані: Тільки transform (y тут не потрібен)
# Також передаємо df_cleaned_final[cat_cols_final], а не просто список назв
cat_preped_final = cat_prep.transform(df_cleaned_final[cat_cols_final])

cat_preped_final.head()
```

```python
num_preped_final = num_data_prep.transform(df_cleaned_final[num_cols_final])
num_preped_final.head()
```

```python
df_final = concat_dfs(cat_preped_final, num_preped_final)

df_final.head()
```

```python
# Отримуємо правильний порядок ознак безпосередньо з навченої моделі (booster)
cols_when_model_builds = final_model.feature_names_in_

# Перевпорядковуємо df_final, щоб стовпці йшли у точній послідовності, яку очікує модель
# Це також автоматично відфільтрує зайві стовпці, якщо такі є
df_final = df_final[cols_when_model_builds]

# Тепер виконуємо передбачення без помилок
prediction_final = final_model.predict(df_final)
```
